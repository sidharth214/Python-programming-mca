```python
#accept 5 students name,rollno,and map these to dictionary and sort it in acsending and descending
students={}
r1=int(input("Enter the roll number:"))
s1=input("Enter the name:")
students[r1]=s1

r2=int(input("Enter the roll number:"))
s2=input("Enter the name:")
students[r2]=s2

r3=int(input("Enter the roll number:"))
s3=input("Enter the name:")
students[r3]=s3

r4=int(input("Enter the roll number:"))
s4=input("Enter the name:")
students[r4]=s4

r5=int(input("Enter the roll number:"))
s5=input("Enter the name:")
students[r5]=s5

print("Student lists:",students)
print("ascending order:",dict(sorted(students.items())))
print("descending order:",dict(sorted(students.items(),reverse=True)))
```
