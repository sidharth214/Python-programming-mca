```python
#to change to a new string where the first and last chars have been exchanged
n = input("Enter the string: ")
print(n[5] + n[1:5] + n[0])
```

```python
#to get a string made of first 2 and last 2 chars from a given string 
n=input("Enter a string : ")
r=n[:2]+ n[-2:]
print(r)
```

```python
#swapping character position 
n1=input("Enter the first  string : ")
n2=input("Enter the second string : ")
r1=n2[0]+n1[1:]
r2=n1[0]+n2[1:]
print(r1,r2)
```

```python
#given string where all occurrences of its first char have been changed to $,except the first char itself 
n=input("Enter a string : ")
r=n[0:1]
s=r+n[1:].replace('o','$')
print(s)
```
