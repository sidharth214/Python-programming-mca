```python
#tuple
#3rd element and 3rd element from last 
a=input("Enter the numbers seperated by commas:")
b=a.split(",")
print("3rd element:",b[2])
print("3rd element from last:",b[-3])
```

```python
#tuple it and find the average
a=input("Enter the numbers seperated by commas:")
b=a.split(",")
b=tuple(int(x)for x in b)
print("tuple:",b)
print("average:",sum(b)/len(b))
```
