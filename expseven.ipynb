```python
#print all colors from colorset1 not containedin colorset2
a=set (input("Enter the first set colors:").split(','))
b=set (input("Enter the second set colors:").split(','))
print("colors are:",a.difference(b))
```

```python
#all colors in set1 and set2
a=set (input("Enter the first set colors:").split(','))
b=set (input("Enter the second set colors:").split(','))
print("colors are:",a.union(b))
```
